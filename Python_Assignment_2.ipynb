```python
#1. Prompt user for Input. Score range should be from 0 to 10 (both inclusive).
#2. Performance Category Program:
```

```python

score = int(input("Enter your score (0-10)"))

if score < 0 or score > 10:
    print("Invalid score. Please enter a score between 0 and 10.")

elif score > 7:
    print("Performance: Above Average")

elif score >= 4 and score <= 7:
    print("Performance: Average")

elif score > 4:
    print("Performance: Below Average")

else:
    print("Performance: Poor")
```

```python
#3. Print the Performance category

print("Performance Categories:")
print("1. Above Average: Score greater than 7")
print("2. Average: Score between 4 and 7 (inclusive)")
print("3. Below Average: Score less than 4")
```

```python
#4. Performance category	Message
# Above Average (>7)	Excellent work!- Keep up the great performance and continue aiming higher.
# Average (4–7)	Good effort! - Keep practising and focus on improving your performance.
```

```python

score = int(input("Enter your score (0 to 10): "))

if score <0 or score > 10:
    print("Invalid score. Please enter a score between 0 and 10.")

elif score > 7:
    print("Performance Category: Above Average")
    print("Excellent work! Keep up the great performance.")

elif score >= 4:
    print("Performance Category: Average")
    print("Good effort! Keep practising to improve your performance.")

else:
   print("Performance Category: Below Average")
   print("Need to improve your performance. Consistent practice will lead to better results.")
```
